# Data Cleaning

```python
school['latest_year_enrolment_FTE']=pd.to_numeric(school['latest_year_enrolment_FTE'], errors='coerce')
school['ICSEA_value']=pd.to_numeric(school['ICSEA_value'], errors='coerce')
school['FOEI_Value']=pd.to_numeric(school['FOEI_Value'], errors='coerce')
school['Indigenous_pct']=pd.to_numeric(school['Indigenous_pct'], errors='coerce')
school['LBOTE_pct']=pd.to_numeric(school['LBOTE_pct'], errors='coerce')
school['Latitude']=pd.to_numeric(school['Latitude'], errors='coerce')
school['Longitude']=pd.to_numeric(school['Longitude'], errors='coerce')
```

```python
school['School_code']= pd.to_numeric(school['School_code'], errors='coerce').astype('Int64')
school['Postcode']= pd.to_numeric(school['Postcode'], errors='coerce').astype('Int64')
```

```python
school.info()
```

```python
school.isnull().sum()
```

```python
columns_drop = [
    'Fax',
    'Support_classes',
    'Operational_directorate_office',
    'Operational_directorate_office_phone',
    'Operational_directorate_office_address',
    'Assets unit',
    'AgeID',
    'Phone',
    'School_Email',
    'Website',
    'electorate_2015_2022',
    'fed_electorate_2016_2024',
    'Late_opening_school',
    'Intensive_english_centre',
    'Date_extracted'
]

school = school.drop(columns=columns_drop)
```

```python
school.shape
```

```python
school.head(1)
```

```python
school.isnull().sum()
```

```python
fill_unknown = [
    'Town_suburb',
    'SA4',
    'FACS_district',
    'Local_health_district',
    'AECG_region'
]

for col in fill_unknown:
    school[col] = school[col].replace('', 'Unknown').fillna('Unknown')
    print(f'{col}: {school[col].isna().sum()} missing remaining')
```

```python
school.isnull().sum()
```

```python
print(f'Duplicate rows: {school.duplicated().sum()}')
print(f'Duplicate School_code: {school["School_code"].duplicated().sum()}')
print(f'Duplicate School_name: {school["School_name"].duplicated().sum()}')
```

```python
# Updated bounds including Lord Howe Island
invalid_lat = school[(school['Latitude'] < -38) | (school['Latitude'] > -28)]
invalid_lon = school[(school['Longitude'] < 141) | (school['Longitude'] > 160)]

print(f'Invalid Latitude: {len(invalid_lat)}')
print(f'Invalid Longitude: {len(invalid_lon)}')
```

```python
cat_cols = [
    'Level_of_schooling', 'Selective_school', 
    'School_gender', 'ASGS_remoteness'
]

for col in cat_cols:
    print(f'\n{col}:')
    print(school[col].value_counts())
```

# Outliers

```python
numeric_cols = [
    'latest_year_enrolment_FTE',
    'ICSEA_value',
    'FOEI_Value',
    'Indigenous_pct',
    'LBOTE_pct'
]

for col in numeric_cols:
    q1    = school[col].quantile(0.25)
    q3    = school[col].quantile(0.75)
    iqr   = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    
    outliers = school[(school[col] < lower) | (school[col] > upper)]
    
    print(f'\n=== {col} ===')
    print(f'  Q1={q1:.1f}, Q3={q3:.1f}, IQR={iqr:.1f}')
    print(f'  Lower fence={lower:.1f}, Upper fence={upper:.1f}')
    print(f'  Outliers below lower: {(school[col] < lower).sum()}')
    print(f'  Outliers above upper: {(school[col] > upper).sum()}')
    print(f'  Total outliers: {len(outliers)}')
```

```python
fig, axes = plt.subplots(1, 5, figsize=(18, 6))

numeric_cols = [
    'latest_year_enrolment_FTE',
    'ICSEA_value',
    'FOEI_Value',
    'Indigenous_pct',
    'LBOTE_pct'
]

for i, col in enumerate(numeric_cols):
    axes[i].boxplot(school[col].dropna(), patch_artist=True)
    axes[i].set_title(col, fontsize=10)
    axes[i].set_ylabel('Value')

plt.suptitle('Boxplots — NSW Public Schools Dataset', fontsize=13)
plt.tight_layout()
plt.show()
```
